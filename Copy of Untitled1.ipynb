```python
import numpy as np
import pandas as pd

# -------------------------------
# 1. Tenure Binning
# -------------------------------
tenure_cutoffs = [0, 12, 24, 48, 60, 100]
period_labels = ['0-1yr', '1-2yr', '2-4yr', '4-5yr', '5+yr']

df['tenure_group'] = pd.cut(df['tenure'], bins=tenure_cutoffs, labels=period_labels)
```

```python
highest_tenure = df['tenure'].max()
df['Recency'] = highest_tenure - df['tenure']

# Frequency (count of active services)
service_cols = [col for col in df.columns if any(x in col for x in [
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
])]

df['Frequency'] = df[service_cols].sum(axis=1)

# Monetary
df['Monetary'] = df['MonthlyCharges']
```

```python
df['usage_ratio'] = df['MonthlyCharges'] / (df['TotalCharges'] + 1)
```

```python
if 'Contract_Month-to-month' in df.columns:
    df['is_monthly_contract'] = df['Contract_Month-to-month']
else:
    df['is_monthly_contract'] = 0

```

```python
np.random.seed(42)

df['support_tickets'] = np.random.randint(0, 5, size=len(df))
df['sentiment_score'] = np.random.uniform(-1, 1, size=len(df))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score, roc_curve, precision_recall_curve,
    classification_report
)
```

```python
X = df.drop('Churn', axis=1)
y = df['Churn']

# -------------------------------
# 2. Train-Test Split
# -------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
model = RandomForestClassifier(
    random_state=42,
    n_jobs=-1
)

# One-hot encode the 'tenure_group' column and align columns for train and test sets
X_train_encoded = pd.get_dummies(X_train, columns=['tenure_group'], drop_first=True)
X_test_encoded = pd.get_dummies(X_test, columns=['tenure_group'], drop_first=True)

# Align columns to handle potential differences in categories between train and test
train_cols = set(X_train_encoded.columns)
test_cols = set(X_test_encoded.columns)

# Add missing columns to the test set, filled with zeros
for col in train_cols - test_cols:
    X_test_encoded[col] = 0

# Add missing columns to the train set, filled with zeros (less likely, but for robustness)
for col in test_cols - train_cols:
    X_train_encoded[col] = 0

# Ensure the order of columns is the same for both datasets
X_test_encoded = X_test_encoded[X_train_encoded.columns]

model.fit(X_train_encoded, y_train)
```

```python
y_pred = model.predict(X_test_encoded)
y_prob = model.predict_proba(X_test_encoded)[:, 1]
```

```python
auc_score = roc_auc_score(y_test, y_prob)

print("AUC-ROC Score:", round(auc_score, 4))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# One-hot encode the 'tenure_group' column in the full X DataFrame for cross-validation
X_encoded_cv = pd.get_dummies(X, columns=['tenure_group'], drop_first=True)

cv_scores = cross_val_score(
    model,
    X_encoded_cv,
    y,
    cv=skf,
    scoring='roc_auc'
)

print("\nCross-validation AUC scores:", cv_scores)
print("Mean CV AUC:", round(cv_scores.mean(), 4))
```

```python
fpr, tpr, _ = roc_curve(y_test, y_prob)

plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {auc_score:.3f}")
plt.plot([0,1], [0,1], linestyle='--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
precision, recall, _ = precision_recall_curve(y_test, y_prob)

plt.figure(figsize=(6,5))
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.grid(alpha=0.3)
plt.show()
```

```python
importances = model.feature_importances_
indices = np.argsort(importances)[-10:]

plt.figure(figsize=(8,5))
plt.barh(range(len(indices)), importances[indices])
plt.yticks(range(len(indices)), X.columns[indices])
plt.title("Top 10 Feature Importances")
plt.show()
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import roc_auc_score
import optuna
```

```python
smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(X_train_encoded, y_train)

print("Before SMOTE:", y_train.value_counts())
print("After SMOTE:", y_train_smote.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_smote, y_train_smote)
```

```python
y_prob = rf_model.predict_proba(X_test_encoded)[:, 1]

auc_score = roc_auc_score(y_test, y_prob)
print("AUC after SMOTE + class weight:", auc_score)
```

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [5, 10, None],
    'min_samples_split': [2, 5]
}

grid_model = GridSearchCV(
    estimator=RandomForestClassifier(class_weight='balanced', random_state=42),
    param_grid=param_grid,
    scoring='roc_auc',
    cv=3,
    n_jobs=-1
)

grid_model.fit(X_train_smote, y_train_smote)

print("Best Parameters:", grid_model.best_params_)
```

```python
best_model = grid_model.best_estimator_

y_prob = best_model.predict_proba(X_test_encoded)[:, 1]

print("Tuned Model AUC:", roc_auc_score(y_test, y_prob))
```

```python
def objective(trial):
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 300),
        'max_depth': trial.suggest_int('max_depth', 5, 20),
        'min_samples_split': trial.suggest_int('min_samples_split', 2, 10)
    }

    model = RandomForestClassifier(
        **params,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train_smote, y_train_smote)
    preds = model.predict_proba(X_test_encoded)[:, 1]

    return roc_auc_score(y_test, preds)

study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=10)

print("Best Optuna Params:", study.best_params)
```

```python
final_model = RandomForestClassifier(
    **study.best_params,
    class_weight='balanced',
    random_state=42
)

final_model.fit(X_train_smote, y_train_smote)

y_prob = final_model.predict_proba(X_test_encoded)[:, 1]

print("Final AUC:", roc_auc_score(y_test, y_prob))
```

```python
import numpy as np
import pandas as pd

# Load data
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

# --- Cleaning ---
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df = df.dropna()
df.drop('customerID', axis=1, inplace=True)
df['Churn'] = df['Churn'].map({'Yes':1, 'No':0})

# Convert categorical → numeric
df = pd.get_dummies(df)

# --- Feature Engineering ---
tenure_cutoffs = [0, 12, 24, 48, 60, 100]
labels = ['0-1yr', '1-2yr', '2-4yr', '4-5yr', '5+yr']
df['tenure_group'] = pd.cut(df['tenure'], bins=tenure_cutoffs, labels=labels)
df = pd.get_dummies(df, columns=['tenure_group'])

# RFM-like features
max_tenure = df['tenure'].max()
df['Recency'] = max_tenure - df['tenure']

service_cols = [col for col in df.columns if 'Yes' in col]
df['Frequency'] = df[service_cols].sum(axis=1)

df['Monetary'] = df['MonthlyCharges']

# Usage ratio
df['usage_ratio'] = df['MonthlyCharges'] / (df['TotalCharges'] + 1)

# Monthly contract flag
df['is_monthly_contract'] = df.filter(regex='Contract_Month-to-month').sum(axis=1)

# Synthetic features
np.random.seed(42)
df['support_tickets'] = np.random.randint(0, 5, size=len(df))
df['sentiment_score'] = np.random.uniform(-1, 1, size=len(df))

# Save final file
df.to_csv("feature_engineered_data.csv", index=False)

print("✅ feature_engineered_data.csv created!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import os

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


def execute_segmentation_pipeline(data_filepath, model_filepath):

    print("🚀 Starting Customer Segmentation Pipeline...\n")

    if not os.path.exists(data_filepath):
        print("❌ Dataset file not found")
        return

    dataset = pd.read_csv(data_filepath)
    print(f"✅ Data Loaded: {dataset.shape}")

    segmentation_features = [
        'Recency',
        'Frequency',
        'Monetary',
        'usage_ratio',
        'support_tickets',
        'sentiment_score'
    ]

    # Ensure all columns exist
    for col in segmentation_features:
        if col not in dataset.columns:
            dataset[col] = 0

    X = dataset[segmentation_features]

    # ------------------------------- #
    # 1. Feature Scaling
    # ------------------------------- #
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    print("✅ Features Scaled")

    # ------------------------------- #
    # 2. Determine Optimal Clusters (Elbow Method and Silhouette Score)
    # ------------------------------- #
    inertia = []
    silhouette_scores = []
    K_range = range(2, 11)  # Typically check K from 2 to 10

    for k in K_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(X_scaled)
        inertia.append(kmeans.inertia_)
        silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))

    # Plot Elbow Method
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(K_range, inertia, marker='o')
    plt.title('Elbow Method For Optimal K')
    plt.xlabel('Number of Clusters (K)')
    plt.ylabel('Inertia')
    plt.grid(True)

    # Plot Silhouette Scores
    plt.subplot(1, 2, 2)
    plt.plot(K_range, silhouette_scores, marker='o')
    plt.title('Silhouette Score For Optimal K')
    plt.xlabel('Number of Clusters (K)')
    plt.ylabel('Silhouette Score')
    plt.grid(True)
    plt.tight_layout()
    plt.show()
    print("✅ Optimal K determination plots generated.")

    # ------------------------------- #
    # 3. Apply K-Means Clustering
    # ------------------------------- #
    # Based on the plots, choose an optimal K. For demonstration, let's pick 3.
    optimal_k = 3 # This should ideally be determined from the plots interactively

    kmeans_final = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
    dataset['Segment'] = kmeans_final.fit_predict(X_scaled)
    print(f"✅ Customers segmented into {optimal_k} clusters.")

    # ------------------------------- #
    # 4. Save Model and Scaler
    # ------------------------------- #
    joblib.dump(kmeans_final, f"{model_filepath}_kmeans.pkl")
    joblib.dump(scaler, f"{model_filepath}_scaler.pkl")
    print("✅ KMeans Model and Scaler Saved.")

    # ------------------------------- #
    # 5. Evaluation
    # ------------------------------- #
    sil_score = silhouette_score(X_scaled, dataset['Segment'])
    print(f"\n✅ Silhouette Score: {sil_score:.4f}")

    # ------------------------------- #
    # 6. Cluster Summary
    # ------------------------------- #
    print("\n📊 Cluster Summary:")
    print(dataset.groupby('Segment')[segmentation_features].mean())

    # ------------------------------- #
    # 7. Visualization
    # ------------------------------- #
    plt.figure()
    sns.scatterplot(
        x=dataset['Recency'],
        y=dataset['Monetary'],
        hue=dataset['Segment']
    )
    plt.title("Customer Segmentation (Recency vs Monetary)")
    plt.show()

    # ------------------------------- #
    # 8. Add Churn Risk
    # ------------------------------- #
    churn_model_path = "finalized_churn_forecaster.pkl" # Assuming this is the saved churn model
    if os.path.exists(churn_model_path):
        print("\n🤖 Loading Churn Prediction Model...")

        churn_model = joblib.load(churn_model_path)

        # Prepare data for churn prediction (ensure feature alignment)
        X_churn = dataset.drop(['Churn', 'Segment'], axis=1, errors='ignore')

        # Need to align columns with the model's expected features
        # This assumes the churn model was trained on the full feature set as available in 'dataset'
        model_features = churn_model.feature_names_in_ if hasattr(churn_model, 'feature_names_in_') else None

        if model_features is not None:
            # Add missing columns to X_churn, fill with 0
            for col in model_features:
                if col not in X_churn.columns:
                    X_churn[col] = 0
            # Drop extra columns from X_churn not in model_features
            X_churn = X_churn[model_features].copy()
        else:
            print("⚠️ Warning: Churn model does not expose feature_names_in_. Attempting prediction with current features.")

        dataset['Churn_Risk'] = churn_model.predict_proba(X_churn)[:, 1]

        print("✅ Churn Risk Added")

    else:
        print("⚠️ Churn prediction model not found, skipping churn prediction.")
        dataset['Churn_Risk'] = np.nan

    # ------------------------------- #
    # 9. Final Insights
    # ------------------------------- #
    final_features = segmentation_features + (['Churn_Risk'] if 'Churn_Risk' in dataset.columns else [])

    print("\n🔥 Final Segment Insights:")
    print(dataset.groupby('Segment')[final_features].mean())

    print("✨ Customer Segmentation Pipeline Completed!")
    dataset.to_csv("final_dataset.csv", index=False)
    print("✅ final_dataset.csv saved successfully!")
    return dataset
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def simulate_retention_strategies(customer_df, risk_col='Churn_Risk', iterations=1000):

    print("🚀 Starting Retention Simulation...\n")

    # -------------------------------
    # SAFETY CHECKS
    # -------------------------------
    if risk_col not in customer_df.columns:
        print(f"❌ Column '{risk_col}' not found in dataset")
        return

    # Work on a copy (safe practice)
    df = customer_df.copy()

    # Handle missing values
    df[risk_col] = df[risk_col].fillna(df[risk_col].mean())

    # Random generator
    rng = np.random.default_rng(42)
    n = len(df)

    # -------------------------------
    # STEP 1: Treatment Assignment
    # -------------------------------
    df['Discount'] = rng.binomial(1, 0.5, n)
    df['Support'] = rng.binomial(1, 0.5, n)
    df['Product'] = rng.binomial(1, 0.5, n)

    # -------------------------------
    # STEP 2: Uplift Simulation
    # -------------------------------
    base_risk = df[risk_col]

    df['Discount_Effect'] = df['Discount'] * (-0.10 * base_risk)
    df['Support_Effect'] = df['Support'] * (-0.15 * base_risk)
    df['Product_Effect'] = df['Product'] * (-0.08 * base_risk)

    total_effect = df[['Discount_Effect', 'Support_Effect', 'Product_Effect']].sum(axis=1)

    df['Adjusted_Risk'] = (base_risk + total_effect).clip(0, 1)

    # -------------------------------
    # STEP 3: MONTE CARLO SIMULATION
    # -------------------------------
    print(f"🔁 Running Monte Carlo ({iterations} iterations)...")

    results = []

    for _ in range(iterations):
        simulated = rng.binomial(1, df['Adjusted_Risk'])
        results.append(simulated.mean())

    before = base_risk.mean()
    after = np.mean(results)
    reduction = before - after

    print("\n📊 RESULTS")
    print(f"Baseline Churn Risk : {before:.4f}")
    print(f"After Intervention  : {after:.4f}")
    print(f"Churn Reduction     : {reduction:.4f}")

    # -------------------------------
    # STEP 4: STRATEGY IMPACT
    # -------------------------------
    impact = {
        "Discount": df['Discount_Effect'].mean(),
        "Support": df['Support_Effect'].mean(),
        "Product": df['Product_Effect'].mean()
    }

    print("\n🔥 Strategy Effectiveness:")
    for k, v in impact.items():
        print(f"{k}: {v:.4f}")

    # -------------------------------
    # STEP 5: VISUALIZATION
    # -------------------------------
    labels = list(impact.keys())
    values = [abs(v) for v in impact.values()]

    plt.figure()
    sns.barplot(x=labels, y=values)
    plt.title("Intervention Impact on Churn Reduction")
    plt.ylabel("Reduction Magnitude")
    plt.xlabel("Strategy")
    plt.show()

    return df, impact, before, after, reduction
```

```python
import mlflow
import mlflow.sklearn
from sklearn.metrics import roc_auc_score, f1_score

mlflow.set_experiment("Churn_Model_Comparison")

def log_model_run(model, model_name, X_test, y_test):

    probs = model.predict_proba(X_test)[:,1]
    preds = model.predict(X_test)

    auc = roc_auc_score(y_test, probs)
    f1 = f1_score(y_test, preds)

    with mlflow.start_run(run_name=model_name):
        mlflow.log_param("model", model_name)
        mlflow.log_metric("AUC", auc)
        mlflow.log_metric("F1_score", f1)

        mlflow.sklearn.log_model(model, "model")

    print(f"✅ Logged {model_name}")
```
